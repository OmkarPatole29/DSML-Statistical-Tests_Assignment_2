```python
# Import libraries for data handling and statistical tests
import pandas as pd
from scipy.stats import ttest_ind, chi2_contingency
```

```python
# Load the Student Performance dataset
df = pd.read_csv("StudentsPerformance.csv")

print(df.head())
print(df.columns)
print(df.shape)
```

```python
# Separate mathematics scores of male and female students
male = df[df["gender"] == "male"]["math score"]
female = df[df["gender"] == "female"]["math score"]

# Compare the average scores using Welch's t-test
t_stat, p_value = ttest_ind(male, female, equal_var=False)

print("Male average:", male.mean())
print("Female average:", female.mean())
print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Significant difference in math scores")
else:
    print("No significant difference in math scores")
```

```python
# Classify students as Pass or Fail based on math scores
df["performance"] = df["math score"].apply(
    lambda score: "Pass" if score >= 50 else "Fail"
)

# Display the first five records
print(df[["parental level of education", "math score", "performance"]].head())
```

```python
# Create a frequency table of parental education and performance
table = pd.crosstab(
    df["parental level of education"],
    df["performance"]
)

# Perform the chi-square test of independence
chi_stat, p_value, dof, expected = chi2_contingency(table)

print("Contingency Table:")
print(table)
print("Chi-square statistic:", chi_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("There is a significant relationship")
else:
    print("No significant relationship found")
```
